```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. CARREGAMENTO DO DATASET
# ------------------------------------------------------------

arquivo = "datatran2025.csv"

df = pd.read_csv(
    arquivo,
    sep=";",
    encoding="latin1",
    low_memory=False
)

print("Dataset carregado com sucesso!")


# ------------------------------------------------------------
# 2. VISÃO GERAL
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("INFORMAÇÕES GERAIS")
print("=" * 60)

print(f"Quantidade de registros: {df.shape[0]:,}")
print(f"Quantidade de colunas:   {df.shape[1]}")

print("\nColunas disponíveis:")
print(df.columns.tolist())

print("\nPrimeiros registros:")
display(df.head())


# ------------------------------------------------------------
# 3. TIPOS DE DADOS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TIPOS DE DADOS")
print("=" * 60)

display(df.dtypes.to_frame("Tipo"))


# ------------------------------------------------------------
# 4. VALORES AUSENTES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("VALORES AUSENTES")
print("=" * 60)

nulos = df.isnull().sum()
nulos = nulos[nulos > 0].sort_values(ascending=False)

if len(nulos) > 0:

    display(nulos.to_frame("Quantidade"))

    plt.figure(figsize=(12, 6))

    nulos.plot(kind="bar")

    plt.title("Quantidade de valores ausentes por variável")
    plt.xlabel("Variável")
    plt.ylabel("Quantidade de valores ausentes")
    plt.xticks(rotation=90)

    plt.tight_layout()
    plt.show()

else:
    print("Não foram encontrados valores ausentes.")


# ------------------------------------------------------------
# 5. ANÁLISE DE ACIDENTES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ANÁLISE DE ACIDENTES")
print("=" * 60)

print(f"Total de registros: {len(df):,}")

col_id = "id" if "id" in df.columns else ("cd_bat" if "cd_bat" in df.columns else None)

if col_id:
    print(f"Acidentes únicos: {df[col_id].nunique():,}")
    print(
        f"Registros além do primeiro por acidente: "
        f"{len(df) - df[col_id].nunique():,}"
    )


# ------------------------------------------------------------
# 6. CLASSIFICAÇÃO DOS ACIDENTES
# ------------------------------------------------------------

if "classificacao_acidente" in df.columns:

    classificacao = df["classificacao_acidente"].value_counts()

    print("\n" + "=" * 60)
    print("CLASSIFICAÇÃO DOS ACIDENTES")
    print("=" * 60)

    display(classificacao.to_frame("Quantidade"))

    plt.figure(figsize=(10, 5))

    classificacao.plot(kind="bar")

    plt.title("Classificação dos acidentes")
    plt.xlabel("Classificação")
    plt.ylabel("Quantidade")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 7. PRINCIPAIS CAUSAS
# ------------------------------------------------------------

col_causa = "causa_acidente" if "causa_acidente" in df.columns else "causa_principal"

if col_causa in df.columns:

    if "causa_principal" in df.columns and col_causa == "causa_acidente":
        df_causas = df[df["causa_principal"] == "Sim"]
    else:
        df_causas = df

    causas = df_causas[col_causa].value_counts().head(10)

    print("\n" + "=" * 60)
    print("10 PRINCIPAIS CAUSAS DE ACIDENTES")
    print("=" * 60)

    display(causas.to_frame("Quantidade"))

    plt.figure(figsize=(10, 6))

    sns.barplot(x=causas.values, y=causas.index, palette="Blues_r")

    plt.title("10 principais causas de acidentes")
    plt.xlabel("Quantidade")
    plt.ylabel("Causa")

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 8. ACIDENTES AO LONGO DO TEMPO
# ------------------------------------------------------------

if "data_inversa" in df.columns:

    df["data_inversa"] = pd.to_datetime(
        df["data_inversa"],
        format="%Y-%m-%d",
        errors="coerce"
    )

    acidentes_mes = (
        df.groupby(df["data_inversa"].dt.to_period("M"))
          .size()
    )

    acidentes_mes.index = acidentes_mes.index.astype(str)

    print("\n" + "=" * 60)
    print("ACIDENTES POR MÊS")
    print("=" * 60)

    display(acidentes_mes.to_frame("Quantidade"))

    plt.figure(figsize=(14, 6))

    acidentes_mes.plot(
        kind="line",
        marker="o"
    )

    plt.title("Quantidade de acidentes ao longo do tempo")
    plt.xlabel("Mês")
    plt.ylabel("Quantidade de acidentes")
    plt.xticks(rotation=45)
    plt.grid()

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 9. ACIDENTES POR FASE DO DIA
# ------------------------------------------------------------

if "fase_dia" in df.columns:

    fase = df["fase_dia"].value_counts()

    print("\n" + "=" * 60)
    print("ACIDENTES POR FASE DO DIA")
    print("=" * 60)

    display(fase.to_frame("Quantidade"))

    plt.figure(figsize=(10, 5))

    fase.plot(kind="bar")

    plt.title("Acidentes por fase do dia")
    plt.xlabel("Fase do dia")
    plt.ylabel("Quantidade")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 10. CONDIÇÕES METEOROLÓGICAS
# ------------------------------------------------------------

col_met = next((c for c in ["condicao_metereologica", "condicao_meteorologica", "cond_meteorologica"] if c in df.columns), None)

if col_met:

    meteorologia = (
        df[col_met]
        .value_counts()
        .head(10)
    )

    print("\n" + "=" * 60)
    print("CONDIÇÕES METEOROLÓGICAS")
    print("=" * 60)

    display(meteorologia.to_frame("Quantidade"))

    plt.figure(figsize=(10, 5))

    meteorologia.plot(kind="bar")

    plt.title("Acidentes por condição meteorológica")
    plt.xlabel("Condição meteorológica")
    plt.ylabel("Quantidade")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 11. RODOVIAS COM MAIS REGISTROS
# ------------------------------------------------------------

col_rod = next((c for c in ["br", "rodovia"] if c in df.columns), None)

if col_rod:

    rodovias = df[col_rod].value_counts().head(10)

    print("\n" + "=" * 60)
    print("10 RODOVIAS COM MAIS REGISTROS")
    print("=" * 60)

    display(rodovias.to_frame("Quantidade"))

    plt.figure(figsize=(10, 5))

    rodovias.plot(kind="bar")

    plt.title("10 rodovias com maior número de registros")
    plt.xlabel("Rodovia / BR")
    plt.ylabel("Quantidade")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 12. RESUMO DA EDA
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("RESUMO DA EDA")
print("=" * 60)

print(f"Registros analisados: {len(df):,}")
print(f"Colunas disponíveis:  {len(df.columns)}")

if col_id:
    print(f"Acidentes únicos:     {df[col_id].nunique():,}")

print(
    f"Valores ausentes:     "
    f"{df.isnull().sum().sum():,}"
)

print("\nEDA concluída com sucesso!")
```

```python
# ------------------------------------------------------------
# 13. CORREÇÃO DE TIPOS E CAMPOS NUMÉRICOS (decimal com vírgula)
# ------------------------------------------------------------

for col in ["latitude", "longitude", "km"]:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.replace(",", ".", regex=False)
        )
        df[col] = pd.to_numeric(df[col], errors="coerce")

# ------------------------------------------------------------
# 14. ATRIBUTOS TEMPORAIS
# ------------------------------------------------------------

df["dia_semana_num"] = df["data_inversa"].dt.dayofweek  # 0=segunda
df["mes"] = df["data_inversa"].dt.month
df["ano"] = df["data_inversa"].dt.year

df["hora"] = pd.to_datetime(
    df["horario"], format="%H:%M:%S", errors="coerce"
).dt.hour

# ------------------------------------------------------------
# 15. CHAVE DE SEGMENTO RODOVIÁRIO (BR + trecho de KM)
# ------------------------------------------------------------

TAMANHO_SEGMENTO_KM = 5  # ajustar depois conforme granularidade desejada

df["km_segmento"] = (df["km"] // TAMANHO_SEGMENTO_KM) * TAMANHO_SEGMENTO_KM
df["segmento_id"] = (
    df["br"].astype(str) + "_KM" + df["km_segmento"].astype(str)
)

print(f"Segmentos únicos identificados: {df['segmento_id'].nunique():,}")
```

```python
# ------------------------------------------------------------
# 16. DEFINIÇÃO DO TARGET
# ------------------------------------------------------------

target = "classificacao_acidente"

print("Distribuição das classes (target):")
display(df[target].value_counts())
display(df[target].value_counts(normalize=True) * 100)

# ------------------------------------------------------------
# 17. REMOÇÃO DE COLUNAS COM VAZAMENTO DE DADOS (data leakage)
# ------------------------------------------------------------

colunas_vazamento = [
    "mortos", "feridos_leves", "feridos_graves", "ilesos",
    "feridos", "pessoas"  # 'pessoas' e 'feridos' correlacionam
                          # fortemente com a gravidade, remover por precaução
]

colunas_vazamento = [c for c in colunas_vazamento if c in df.columns]
print(f"Colunas removidas por vazamento: {colunas_vazamento}")

# ------------------------------------------------------------
# 18. SELEÇÃO DE FEATURES
# ------------------------------------------------------------

features_categoricas = [
    "causa_acidente", "tipo_acidente", "fase_dia",
    "condicao_metereologica", "condicao_meteorologica",
    "tipo_pista", "tracado_via", "uso_solo", "sentido_via", "uf"
]
features_categoricas = [c for c in features_categoricas if c in df.columns]

features_numericas = [
    "hora", "mes", "dia_semana_num", "veiculos",
    "latitude", "longitude"
]
features_numericas = [c for c in features_numericas if c in df.columns]

print(f"Features categóricas: {features_categoricas}")
print(f"Features numéricas:   {features_numericas}")

colunas_modelo = features_categoricas + features_numericas + [target]
df_modelo = df[colunas_modelo].copy()

# Remove linhas sem target ou com dados essenciais ausentes
df_modelo = df_modelo.dropna(subset=[target])
print(f"\nRegistros disponíveis para treino: {len(df_modelo):,}")
```

```python
# ------------------------------------------------------------
# 19. PRÉ-PROCESSAMENTO (encoding + split)
# ------------------------------------------------------------

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

X = df_modelo[features_categoricas + features_numericas]
y = df_modelo[target]

preprocessador = ColumnTransformer(
    transformers=[
        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]), features_categoricas),
        ("num", SimpleImputer(strategy="median"), features_numericas)
    ]
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,       # mantém a proporção das classes (fatais são raros)
    random_state=42
)

print(f"Treino: {X_train.shape[0]:,} | Teste: {X_test.shape[0]:,}")
```

```python
# ------------------------------------------------------------
# 20. TREINAMENTO DO MODELO (Random Forest com balanceamento)
# ------------------------------------------------------------

from sklearn.ensemble import RandomForestClassifier

modelo = Pipeline([
    ("preprocessador", preprocessador),
    ("classificador", RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    ))
])

modelo.fit(X_train, y_train)
print("Modelo treinado com sucesso!")
```

```python
# ------------------------------------------------------------
# 21. AVALIAÇÃO DO MODELO
# ------------------------------------------------------------

from sklearn.metrics import (
    classification_report, confusion_matrix,
    ConfusionMatrixDisplay, accuracy_score
)

y_pred = modelo.predict(X_test)

print(f"Acurácia geral: {accuracy_score(y_test, y_pred):.2%}\n")
print("Relatório de classificação:")
print(classification_report(y_test, y_pred))

plt.figure(figsize=(8, 6))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred,
    xticks_rotation=45,
    cmap="Blues"
)
plt.title("Matriz de Confusão - Gravidade do Acidente")
plt.tight_layout()
plt.show()
```

```python
# ------------------------------------------------------------
# 22. INCLUIR SEGMENTO_ID E FAZER O SPLIT ANTES DE CALCULAR RISCO
# ------------------------------------------------------------

colunas_modelo = features_categoricas + features_numericas + ["segmento_id", target]
df_modelo = df[colunas_modelo].copy()
df_modelo = df_modelo.dropna(subset=[target])

X = df_modelo[features_categoricas + features_numericas + ["segmento_id"]]
y = df_modelo[target]

# Split ANTES de calcular qualquer estatística agregada por segmento
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print(f"Treino: {X_train.shape[0]:,} | Teste: {X_test.shape[0]:,}")

# ------------------------------------------------------------
# 23. CALCULAR RISCO HISTÓRICO DO SEGMENTO (usando só o treino)
# ------------------------------------------------------------

treino_temp = X_train.copy()
treino_temp["target_temp"] = y_train.values

risco_segmento = (
    treino_temp
    .assign(fatal=lambda d: (d["target_temp"] == "Com Vítimas Fatais").astype(int))
    .groupby("segmento_id")
    .agg(
        total_acidentes_segmento=("fatal", "count"),
        taxa_fatalidade_segmento=("fatal", "mean")
    )
)

media_global_fatalidade = treino_temp["target_temp"].eq("Com Vítimas Fatais").mean()

def aplicar_risco_segmento(X_split):
    X_novo = X_split.merge(risco_segmento, on="segmento_id", how="left")
    # Segmentos que só aparecem no teste (não vistos no treino) recebem a média global
    X_novo["taxa_fatalidade_segmento"] = X_novo["taxa_fatalidade_segmento"].fillna(media_global_fatalidade)
    X_novo["total_acidentes_segmento"] = X_novo["total_acidentes_segmento"].fillna(0)
    return X_novo.drop(columns=["segmento_id"])

X_train = aplicar_risco_segmento(X_train)
X_test = aplicar_risco_segmento(X_test)

features_numericas_v2 = features_numericas + ["total_acidentes_segmento", "taxa_fatalidade_segmento"]

print("Novas colunas numéricas:", features_numericas_v2)
display(X_train.head())
```

```python
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

!pip install -q imbalanced-learn

# ------------------------------------------------------------
# 24. PIPELINE COM OVERSAMPLING (imblearn) + FEATURE NOVA
# ------------------------------------------------------------

from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import RandomOverSampler
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
import numpy as np # Adicionado para np.round

preprocessador_v2 = ColumnTransformer(
    transformers=[
        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)) # Adicionado sparse_output=False
        ]), features_categoricas),
        ("num", SimpleImputer(strategy="median"), features_numericas_v2)
    ]
)

# Calcular a distribuição das classes no conjunto de treino
class_counts = y_train.value_counts()
majority_class_name = class_counts.idxmax()
majority_class_count = class_counts.max()

# Definir a estratégia de amostragem como um dicionário para multi-class
# Oversample minority classes to 50% of the majority class count
sampling_strategy_dict = {
    cls: int(np.round(majority_class_count * 0.5)) if cls != majority_class_name else majority_class_count
    for cls in class_counts.index
}

modelo_v2 = ImbPipeline([
    ("preprocessador", preprocessador_v2),
    ("oversampling", RandomOverSampler(sampling_strategy=sampling_strategy_dict, random_state=42)),
    ("classificador", RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced_subsample",  # o ajuste 1 que você já fez
        random_state=42,
        n_jobs=-1
    ))
])

modelo_v2.fit(X_train, y_train)
print("Modelo v2 (com oversampling + risco de segmento) treinado!")
```

```python
y_pred_v2 = modelo_v2.predict(X_test)

print(f"Acurácia geral: {accuracy_score(y_test, y_pred_v2):.2%}\n")
print(classification_report(y_test, y_pred_v2))

plt.figure(figsize=(8, 6))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_v2, xticks_rotation=45, cmap="Blues")
plt.title("Matriz de Confusão v2 - com oversampling e risco de segmento")
plt.tight_layout()
plt.show()
```

```python
# ------------------------------------------------------------
# 25. ALTERNATIVA: HISTGRADIENTBOOSTING
# ------------------------------------------------------------

from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.preprocessing import FunctionTransformer

modelo_v3 = ImbPipeline([
    ("preprocessador", preprocessador_v2),
    ("oversampling", RandomOverSampler(random_state=42)),
    ("to_dense", FunctionTransformer(lambda x: x.toarray() if hasattr(x, 'toarray') else x, accept_sparse=True)), # Garante que o HistGradientBoostingClassifier receba dados densos
    ("classificador", HistGradientBoostingClassifier(
        max_iter=300,
        random_state=42
    ))
])

modelo_v3.fit(X_train, y_train)
y_pred_v3 = modelo_v3.predict(X_test)

print(f"Acurácia geral: {accuracy_score(y_test, y_pred_v3):.2%}\n")
print(classification_report(y_test, y_pred_v3))
```

```python
# ------------------------------------------------------------
# 26. REFORMULAÇÃO DO TARGET: BINÁRIO (FATAL vs NÃO-FATAL)
# ------------------------------------------------------------

# Cria o novo target binário a partir da classificação original
df_modelo["target_binario"] = (
    df_modelo[target] == "Com Vítimas Fatais"
).astype(int)  # 1 = fatal, 0 = não-fatal

print("Distribuição do novo target binário:")
print(df_modelo["target_binario"].value_counts())
print(df_modelo["target_binario"].value_counts(normalize=True) * 100)

# Refaz X e y com o novo target
X = df_modelo[features_categoricas + features_numericas + ["segmento_id"]]
y = df_modelo["target_binario"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```
